```python
import pandas as pd
import os
```

```python
annotators = ["A1", "A2", "A3", "A4", "A5"]
for ann in annotators:
    ann_total = 0
    dir = f"{ann}"
    for root, dirs, files in os.walk(dir):
        for file in files:
            if file.endswith(".csv"):
                df = pd.read_csv(os.path.join(root, file))
                ann_total += len(df)
    print(f"{ann}: {ann_total} comments")
```

```python
# Count total number of unique annotated comments
unique_comments = set()
for ann in annotators:
    dir = f"{ann}"
    for root, dirs, files in os.walk(dir):
        for file in files:
            if file.endswith(".csv"):
                df = pd.read_csv(os.path.join(root, file))
                unique_comments.update(df['comment_id'].unique())
print(f"Total unique annotated comments: {len(unique_comments)}")
```
