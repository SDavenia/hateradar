```python
import os
import pandas as pd
```

```python
INPUT_DIR = "VideosComments/youtube/comments_anonymized/{newspaper}"
```

```python
newspapers = ["corriere_della_sera", "repubblica", "il_gazzettino", "ilmessaggero", "lastampa"]
```

```python
num_comments = {newspaper: [] for newspaper in newspapers}
average_comment_tree_depth = {newspaper: [] for newspaper in newspapers}
longest_comment_depth = {newspaper: [] for newspaper in newspapers}
author_comments_newspaper = {newspaper: {} for newspaper in newspapers} # corriere : {author1: num_comments, author2: num_comments, ...}

for newspaper in newspapers:
    for file_path in os.listdir(INPUT_DIR.format(newspaper=newspaper)):
        if file_path.endswith(".csv"):
            video_id = file_path.split(".csv")[0]
            input_file_path = os.path.join(INPUT_DIR.format(newspaper=newspaper), file_path)
            output_file_path = os.path.join(INPUT_DIR.format(newspaper=newspaper), f"{video_id}_cleaned.csv")
            
            # Load the CSV file into a DataFrame
            df = pd.read_csv(input_file_path)
            if df.empty:
                num_comments[newspaper].append(0)
                average_comment_tree_depth[newspaper].append(0)
                longest_comment_depth[newspaper].append(0)
                continue

            # Count the number of comments
            num_comments[newspaper].append(len(df))

            # Calculate the average comment tree depth 
            average_comment_tree_depth[newspaper].append(df["depth"].mean())
            # Calculate the longest comment depth
            longest_comment_depth[newspaper].append(df["depth"].max())

            # Count comments by author
            for author in df["author"].unique():
                if author not in author_comments_newspaper[newspaper]:
                    author_comments_newspaper[newspaper][author] = 0
                author_comments_newspaper[newspaper][author] += df[df["author"] == author].shape[0]
```

```python
# For each dataset print mean num comments, average comment tree depth
print(f"Total number of videos: {sum(len(os.listdir(INPUT_DIR.format(newspaper=newspaper))) for newspaper in newspapers)}")
print(f"Total number of comments: {sum(sum(num_comments[newspaper]) for newspaper in newspapers)}")
for newspaper in newspapers:
    print(f"Newspaper: {newspaper}")
    print(f"Total number of videos: {len(os.listdir(INPUT_DIR.format(newspaper=newspaper)))}")
    print(f"Total number of comments: {sum(num_comments[newspaper])}")
    print(f"Mean number of comments: {sum(num_comments[newspaper]) / len(num_comments[newspaper])}")
    print(f"Average comment tree depth: {sum(average_comment_tree_depth[newspaper]) / len(average_comment_tree_depth[newspaper])}")
    print(f"Longest comment depth: {max(longest_comment_depth[newspaper])}")
    print(f"Percentage of videos with no comments: {num_comments[newspaper].count(0) / len(num_comments[newspaper]) * 100:.2f}%")
    print(f"Top 5 most active authors: {sorted(author_comments_newspaper[newspaper].items(), key=lambda x: x[1], reverse=True)[:5]}")
    print("\n")
```
